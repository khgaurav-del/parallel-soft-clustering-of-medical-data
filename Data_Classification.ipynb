# Perscription Drugs Classification

we will first sort the many different drugs into specific categories, this helps in making better clusters when data is more similar. \
it aslo reduces the number or types of drug, to smaller number. Most patients are prescribed diffrerent drugs with similar affects

```python
import pandas as pd

# Load the data
df = pd.read_csv('MIMIC-III Dataset\MIMIC -III (10000 patients)\PRESCRIPTIONS\PRESCRIPTIONS_sorted.csv', low_memory=False)

def classify_drug_final(drug_name):
    if pd.isna(drug_name):
        return 'Unknown'
    
    d = str(drug_name).lower()
    
    # 0. Medical Supplies / Non-Drugs (Syringes, Vials, Soln)
    if any(x in d for x in ['syringe', 'vial', 'sterile water', 'flush', 'bag', 'sw ', 'soln.']):
        return 'Medical Supplies / Diluents'
    if d in ['soln.', 'soln', 'sw', 'base', 'ointment']:
        return 'Medical Supplies / Diluents'

    # 1. Analgesics
    if any(x in d for x in ['morphine', 'fentanyl', 'hydromorphone', 'oxycodone', 'methadone', 'buprenorphine', 'meperidine', 'remifentanil', 'tramadol']):
        return 'Opioid Analgesics (Severe Pain)'
    if 'acetaminophen' in d and any(x in d for x in ['oxycodone', 'hydrocodone', 'codeine']):
        return 'Opioid/Non-Opioid Combo Analgesics'
    if 'acetaminophen' in d or 'paracetamol' in d or 'tylenol' in d:
        return 'Non-Opioid Analgesics/Antipyretics'
    if any(x in d for x in ['aspirin', 'ibuprofen', 'naproxen', 'ketorolac', 'diclofenac', 'indomethacin', 'celecoxib', 'meloxicam']):
        return 'NSAIDs/Antiplatelets'
    
    # 2. Cardiovascular
    if any(x in d for x in ['metoprolol', 'atenolol', 'propranolol', 'carvedilol', 'labetalol', 'esmolol', 'sotalol', 'nadolol']):
        return 'Beta-Blockers'
    if any(x in d for x in ['lisinopril', 'captopril', 'enalapril', 'ramipril']):
        return 'ACE Inhibitors'
    if any(x in d for x in ['losartan', 'valsartan', 'irbesartan']):
        return 'ARBs (Angiotensin Receptor Blockers)'
    if any(x in d for x in ['amiodarone', 'diltiazem', 'verapamil', 'digoxin', 'lidocaine']):
        return 'Antiarrhythmics'
    if any(x in d for x in ['nitroglycerin', 'isosorbide', 'nitroprusside', 'hydralazine', 'nesiritide']):
        return 'Nitrates/Vasodilators'
    if any(x in d for x in ['amlodipine', 'nifedipine', 'nicardipine', 'nimodipine']):
        return 'Calcium Channel Blockers'
    if any(x in d for x in ['phenylephrine', 'norepinephrine', 'epinephrine', 'dopamine', 'dobutamine', 'vasopressin', 'midodrine']):
        return 'Vasopressors/Inotropes (Shock/Hypotension)'
    if any(x in d for x in ['atorvastatin', 'simvastatin', 'pravastatin', 'rosuvastatin']):
        return 'Statins (Lipid-Lowering)'
    if 'clonidine' in d:
        return 'Alpha-2 Agonists (BP/Sedation)'
    if 'tamsulosin' in d:
        return 'Alpha-1 Blockers'

    # 3. Diuretics
    if any(x in d for x in ['furosemide', 'torsemide', 'bumetanide']):
        return 'Loop Diuretics'
    if 'spironolactone' in d:
        return 'Potassium-Sparing Diuretics'
    if any(x in d for x in ['hydrochlorothiazide', 'chlorothiazide', 'metolazone']):
        return 'Thiazide Diuretics'
    if 'acetazolamide' in d:
        return 'Carbonic Anhydrase Inhibitors'
    if 'mannitol' in d:
        return 'Osmotic Diuretics (ICP Management)'
    
    # 4. Fluids & Electrolytes
    if any(x in d for x in ['d5w', 'dextrose', 'd5', 'd10', 'd50']):
        return 'IV Fluids (Dextrose based)'
    if any(x in d for x in ['ns', 'sodium chloride', 'nacl', '0.9%', 'saline', '0.45%']):
        return 'IV Fluids (Saline based)'
    if 'lr' in d or 'lactated' in d:
        return 'IV Fluids (Lactated Ringers)'
    if any(x in d for x in ['potassium chloride', 'kcl', 'potassium phosphate', 'neutra-phos', 'sodium phosphate']):
        return 'Electrolytes (Potassium/Phosphate)'
    if 'magnesium' in d and 'milk' not in d: # Exclude Milk of Magnesia
        return 'Electrolytes (Magnesium)'
    if 'calcium' in d:
        return 'Electrolytes (Calcium)'
    if 'bicarbonate' in d:
        return 'Electrolytes (Bicarbonate)'
    if 'albumin' in d or 'hespan' in d or 'voluven' in d or 'fat emulsion' in d or 'lipids' in d or 'tpn' in d or 'nutrition' in d or 'mct oil' in d:
        return 'Volume Expanders / TPN (Nutrition)'

    # 5. Endocrine / Metabolic
    if 'insulin' in d:
        return 'Insulins (Glycemic Control)'
    if any(x in d for x in ['metformin', 'glipizide', 'glyburide', 'sitagliptin']):
        return 'Oral Hypoglycemics'
    if any(x in d for x in ['levothyroxine', 'liothyronine']):
        return 'Thyroid Hormones'
    if any(x in d for x in ['calcitriol', 'ergocalciferol', 'cholecalciferol', 'vitamin d', 'fludrocortisone']):
        return 'Steroids / Vitamin D Analogs'
    
    # 6. Gastrointestinal
    if any(x in d for x in ['pantoprazole', 'omeprazole', 'lansoprazole', 'esomeprazole']):
        return 'PPIs (Gastric Acid Reducers)'
    if any(x in d for x in ['ranitidine', 'famotidine', 'cimetidine']):
        return 'H2 Antagonists'
    if any(x in d for x in ['docusate', 'bisacodyl', 'senna', 'polyethylene glycol', 'lactulose', 'miralax', 'milk of magnesia', 'magnesium citrate', 'golytely']):
        return 'Laxatives/Stool Softeners'
    if any(x in d for x in ['ondansetron', 'metoclopramide', 'promethazine', 'prochlorperazine', 'dolasetron', 'granisetron']):
        return 'Antiemetics (Nausea/Vomiting)'
    if 'sucralfate' in d:
        return 'GI Protectants'
    if 'sodium polystyrene' in d or 'kayexalate' in d or 'sevelamer' in d:
        return 'Potassium/Phosphate Binders'
    if 'simethicone' in d:
        return 'Anti-Gas/Anti-flatulents'
    if 'octreotide' in d:
        return 'GI Bleed/Hormone Suppressants'
    if 'loperamide' in d:
        return 'Antidiarrheals'
    
    # 7. Anticoagulants & Hematologic
    if 'heparin' in d or 'enoxaparin' in d or 'dalteparin' in d or 'fondaparinux' in d or 'argatroban' in d or 'bivalirudin' in d:
        return 'Injectable Anticoagulants'
    if 'warfarin' in d or 'apixaban' in d or 'rivaroxaban' in d or 'dabigatran' in d:
        return 'Oral Anticoagulants'
    if 'clopidogrel' in d or 'ticagrelor' in d or 'prasugrel' in d or 'eptifibatide' in d:
        return 'Antiplatelet Agents (Non-Aspirin)'
    if 'alteplase' in d or 'tpa' in d:
        return 'Thrombolytics (Clot Busters)'
    if 'epoetin' in d or 'darbepoetin' in d:
        return 'Erythropoiesis-Stimulating Agents'
    if 'protamine' in d:
        return 'Anticoagulant Reversals'

    # 8. Sedatives, Psychotropics, Anticonvulsants
    if any(x in d for x in ['lorazepam', 'midazolam', 'diazepam', 'alprazolam', 'clonazepam', 'chlordiazepoxide', 'oxazepam']):
        return 'Benzodiazepines (Anxiolytics/Sedatives)'
    if 'propofol' in d or 'dexmedetomidine' in d or 'ketamine' in d:
        return 'General Anesthetics/ICU Sedatives'
    if any(x in d for x in ['haloperidol', 'quetiapine', 'olanzapine', 'risperidone', 'ziprasidone', 'aripiprazole']):
        return 'Antipsychotics'
    if any(x in d for x in ['fluoxetine', 'sertraline', 'citalopram', 'escitalopram', 'trazodone', 'mirtazapine', 'venlafaxine', 'duloxetine', 'bupropion', 'paroxetine']):
        return 'Antidepressants'
    if any(x in d for x in ['zolpidem', 'eszopiclone', 'zaleplon']):
        return 'Sleep Aids / Hypnotics'
    if any(x in d for x in ['gabapentin', 'pregabalin']):
        return 'Neuropathic Pain/Anticonvulsants'
    if any(x in d for x in ['levetiracetam', 'phenytoin', 'valproic acid', 'carbamazepine', 'lacosamide', 'topiramate']):
        return 'Antiepileptics'

    # 9. Antibiotics / Anti-infectives
    if any(x in d for x in ['vancomycin', 'linezolid']):
        return 'Glycopeptide/Oxazolidinone Antibiotics (MRSA coverage)'
    if any(x in d for x in ['levofloxacin', 'ciprofloxacin', 'moxifloxacin']):
        return 'Fluoroquinolone Antibiotics'
    if any(x in d for x in ['ceftriaxone', 'cefepime', 'cefazolin', 'ceftazidime', 'cephalexin', 'cefuroxime']):
        return 'Cephalosporin Antibiotics'
    if any(x in d for x in ['piperacillin', 'tazobactam', 'ampicillin', 'amoxicillin', 'penicillin', 'nafcillin', 'oxacillin']):
        return 'Penicillin/Beta-lactam Antibiotics'
    if any(x in d for x in ['meropenem', 'imipenem', 'ertapenem']):
        return 'Carbapenem Antibiotics'
    if 'metronidazole' in d or 'clindamycin' in d:
        return 'Anaerobic Antibiotics'
    if 'fluconazole' in d or 'micafungin' in d or 'nystatin' in d or 'clotrimazole' in d or 'miconazole' in d:
        return 'Antifungals'
    if any(x in d for x in ['gentamicin', 'tobramycin', 'amikacin']):
        return 'Aminoglycoside Antibiotics'
    if any(x in d for x in ['azithromycin', 'clarithromycin', 'erythromycin']):
        return 'Macrolide Antibiotics'
    if 'sulfamethoxazole' in d or 'trimethoprim' in d or 'bactrim' in d:
        return 'Sulfonamide Antibiotics'
    if 'acyclovir' in d or 'valacyclovir' in d or 'ganciclovir' in d:
        return 'Antivirals'

    # 10. Respiratory & Allergy
    if 'albuterol' in d or 'levalbuterol' in d:
        return 'Beta-2 Agonist Bronchodilators'
    if 'ipratropium' in d or 'tiotropium' in d:
        return 'Anticholinergic Bronchodilators'
    if any(x in d for x in ['fluticasone', 'budesonide', 'methylprednisolone', 'prednisone', 'dexamethasone', 'hydrocortisone']):
        return 'Corticosteroids (Anti-inflammatory)'
    if any(x in d for x in ['diphenhydramine', 'cetirizine', 'loratadine', 'fexofenadine', 'hydroxyzine']):
        return 'Antihistamines'
    if 'acetylcysteine' in d or 'guaifenesin' in d:
        return 'Mucolytics / Expectorants / Antidotes'
    if 'nicotine' in d:
        return 'Smoking Cessation / Nicotine'

    # 11. Immunology, Neuro & Other Specifics
    if any(x in d for x in ['tacrolimus', 'cyclosporine', 'mycophenolate', 'sirolimus']):
        return 'Immunosuppressants (Transplant/Autoimmune)'
    if any(x in d for x in ['phytonadione', 'folic acid', 'thiamine', 'ferrous sulfate', 'multivitamin', 'ascorbic acid', 'cyanocobalamin', 'pyridoxine', 'iron', 'nephrocaps', 'vitamin', 'zinc']):
        return 'Vitamins / Minerals / Supplements'
    if any(x in d for x in ['glycopyrrolate', 'atropine', 'scopolamine']):
        return 'Anticholinergics / Antimuscarinics'
    if 'neostigmine' in d or 'physostigmine' in d:
        return 'Cholinesterase Inhibitors / Reversals'
    if 'cosyntropin' in d:
        return 'Diagnostic Hormones'
    if 'glucagon' in d:
        return 'Hypoglycemia Antidotes'
    if 'caffeine citrate' in d:
        return 'Neonatal Stimulants'
    if 'cisatracurium' in d or 'rocuronium' in d or 'vecuronium' in d:
        return 'Neuromuscular Blockers (Paralytics)'
    if 'vaccine' in d or 'vac polyvalent' in d:
        return 'Vaccines'
    if 'allopurinol' in d:
        return 'Gout/Hyperuricemia Agents'
    if 'tears' in d or 'lotion' in d or 'cepacol' in d or 'ointment' in d:
        return 'Topicals / Eye Drops / Lozenges'

    # Catch-all
    return 'Other / Miscellaneous'

# Apply mapping to Column H ('DRUG')
df['DRUG'] = df['DRUG'].apply(classify_drug_final)

# Make sure we also hit the duplicate DRUG column at the end
if 'DRUG.1' in df.columns:
    df['DRUG.1'] = df['DRUG.1'].apply(classify_drug_final)

df.to_csv('MIMIC-III Dataset\\MIMIC -III (10000 patients)\\PRESCRIPTIONS\\PRESCRIPTIONS_grouped_classes_final.csv', index=False)
```

# Patient Cluster Labelling

this uses the cpt events files and the different categories assigned to the by the hospital administration.\
we are using this to group similar patient types such as "consultancy" and "follow-up checkups". \
the end result is 13 unique categories or clusters with each patient that belongs to the category getting assinged to one of 13 clusters.


```python
import pandas as pd

# Load CPTEVENTS
cpt_df = pd.read_csv(r'MIMIC-III Dataset\MIMIC -III (10000 patients)\CPTEVENTS\CPTEVENTS_sorted.csv', low_memory=False) # Adjust the path as needed

def group_cpt_category(cat):

    if pd.isna(cat):
        return 'Unknown'

    c = str(cat).lower().strip()

    if c == '0':
        return 'Unknown'

    # -----------------------------
    # Critical care / inpatient
    # -----------------------------
    if any(x in c for x in [
        'critical care',
        'hospital inpatient',
        'hospital observation',
        'prolonged services',
        'office/other outpatient'
    ]):
        return 'Inpatient & Critical Care'

    # -----------------------------
    # Consultations / case review
    # -----------------------------
    if any(x in c for x in [
        'consultation',
        'case management',
        'team conference'
    ]):
        return 'Consultations'

    # -----------------------------
    # Cardiovascular
    # -----------------------------
    if any(x in c for x in [
        'cardiovascular',
        'vascular'
    ]):
        return 'Cardiovascular'

    # -----------------------------
    # Pulmonary / Respiratory
    # -----------------------------
    if any(x in c for x in [
        'pulmonary',
        'respiratory',
        'mediastinum',
        'diaphragm'
    ]):
        return 'Pulmonary & Respiratory'

    # -----------------------------
    # Dialysis / Renal
    # -----------------------------
    if any(x in c for x in [
        'dialysis',
        'renal',
        'end-stage renal',
        'urinary system'
    ]):
        return 'Dialysis & Renal'

    # -----------------------------
    # Nervous system / mental health
    # -----------------------------
    if any(x in c for x in [
        'nervous system',
        'neurology',
        'neuromuscular',
        'psychiatry',
        'neuro',
        'mental'
    ]):
        return 'Nervous System & Mental Health'

    # -----------------------------
    # Gastrointestinal
    # -----------------------------
    if any(x in c for x in [
        'digestive'
    ]):
        return 'Gastrointestinal'

    # -----------------------------
    # Musculoskeletal
    # -----------------------------
    if any(x in c for x in [
        'musculoskeletal',
        'bone',
        'joint'
    ]):
        return 'Musculoskeletal'

    # -----------------------------
    # Genitourinary / maternity
    # -----------------------------
    if any(x in c for x in [
        'genital',
        'maternity'
    ]):
        return 'Genitourinary'

    # -----------------------------
    # Integumentary
    # -----------------------------
    if 'integumentary' in c:
        return 'Integumentary'

    # -----------------------------
    # Hematology / oncology / endocrine / labs
    # -----------------------------
    if any(x in c for x in [
        'hemic',
        'hematology',
        'lymphatic',
        'endocrine',
        'chemistry',
        'chemotherapy'
    ]):
        return 'Hem/Onc & Endocrine'

    # -----------------------------
    # Imaging / radiology
    # -----------------------------
    if any(x in c for x in [
        'ultrasound',
        'imaging',
        'radiologic',
        'radiologic guidance'
    ]):
        return 'Imaging & Radiology'

    # -----------------------------
    # Sedation / anesthesia
    # -----------------------------
    if any(x in c for x in [
        'sedation'
    ]):
        return 'Sedation / Anesthesia'

    # -----------------------------
    # Ophthalmology
    # -----------------------------
    if any(x in c for x in [
        'ophthalmology',
        'eye',
        'ocular'
    ]):
        return 'Ophthalmology'

    # -----------------------------
    # Physical medicine / rehab
    # -----------------------------
    if any(x in c for x in [
        'rehabilitation',
        'physical medicine'
    ]):
        return 'Physical Medicine & Rehab'

    # -----------------------------
    # Drug administration / infusions
    # -----------------------------
    if any(x in c for x in [
        'injections',
        'infusions',
        'hydration'
    ]):
        return 'Drug Administration'

    return 'Other / Miscellaneous'


# Apply grouping
cpt_df['MACRO_CATEGORY'] = cpt_df['SUBSECTIONHEADER'].apply(group_cpt_category)


# Remove unknown before dominant calculation
filtered_cpt = cpt_df[cpt_df['MACRO_CATEGORY'] != 'Unknown']


# Dominant CPT category per patient
patient_dominant = filtered_cpt.groupby('HADM_ID')['MACRO_CATEGORY'].apply(
    lambda x: x.mode()[0] if not x.mode().empty else 'Unknown'
).reset_index()


patient_dominant.rename(columns={
    'MACRO_CATEGORY': 'DOMINANT_CPT_CLUSTER'
}, inplace=True)


# Create a numeric ID for each cluster using pd.factorize
# This assigns a unique integer (0, 1, 2...) to each unique cluster string
patient_dominant['DOMINANT_CPT_CLUSTER_ID'] = pd.factorize(patient_dominant['DOMINANT_CPT_CLUSTER'])[0]

# Save the final dataset
patient_dominant.to_csv('PATIENT_DOMINANT_CLUSTERS.csv', index=False)

# Optional: Print out the mapping so you can reference it later
mapping = dict(enumerate(patient_dominant['DOMINANT_CPT_CLUSTER'].unique()))
print("Numeric Map:")
for k, v in mapping.items():
    print(f"{k}: {v}")

# Save
patient_dominant.to_csv(r'MIMIC-III Dataset\MIMIC -III (10000 patients)\CPTEVENTS\PATIENT_DOMINANT_CLUSTERS.csv', index=False) # Adjust the path as needed
```
